```python
# 1 - Demo of function
df_splitter = df['Age']
columnName = 'Earnings_1978'
Threshold = 30
BelowResults = []
AboveResults = []
def age_splitter(df_splitter, columnName, Threshold):
    if columnName in df.columns:
        below_df = df[df_splitter <  Threshold]
        above_df = df[df_splitter >= Threshold] 
        BelowResults.append(below_df[columnName])
        AboveResults.append(above_df[columnName])

    return BelowResults, AboveResults

result = age_splitter(df_splitter, columnName, Threshold) 

print(result)


# 2 Percentage Under 30
below_df = df[df_splitter <  Threshold]
above_df = df[df_splitter >= Threshold] 
print("how many people were below 30:", below_df[columnName].count())
print("how many people were in the dataset:", df_splitter.count())
print("percentage of those under 30:", round((below_df[columnName].count()*100/df_splitter.count()),2), "%")

# 3 - 1978 Earnings Comparison
print("mean for Earnings_1978 of those below 30:", round(below_df[columnName].mean()))
print("mean for Earnings_1978 of people 30 and above:", round(above_df[columnName].mean()))
print(" ")
print("standard deviation for Earnings_1978 of those below 30:", round(below_df[columnName].std()))
print("standard deviation for Earnings_1978 of people 30 and above:", round(above_df[columnName].std()))

sns.lineplot(data=below_df, x="Age", y="Earnings_1978")
sns.lineplot(data=above_df, x="Age", y="Earnings_1978")

'''
# Note the higher standard deviation and thus the wider shade of the older group's lineplot. 
# While the younger group's is characterized by a steeper increase in earnings with age, 
the older group's has more ups and downs in earnings with age. At about 50, it steadily declines. 
'''
```

<ol>
<li> Create a function in the .py file called effectSizer that takes in a dataframe, a numerical column name, a column name of a categorical value that is binary (two values only), and returns a dictionary of the categorical classes and their corresponding effect sizes on the numerical value. </li>
<li> For 1978, which of Race, Hisp, MaritalStatus have the largest effect size. (Use Yes/True/1 for x1)</li>
</ol>

```python
# 1. Demo of function
columnBinCategorical = ['Race','Hisp', 'MaritalStatus', 'Nodeg'] # must be binary, so Eduacation is excluded
columnNumerical = ['Age', 'Earnings_1974', 'Earnings_1975', 'Earnings_1978']

splitterColumn = 'Nodeg'
columnNameNum = 'Earnings_1978' 
Present = []
Absent = []

def effectSizer(df, splitterColumn, columnNameNum): # this function also returns the count for the categorical classes.
    if splitterColumn == 'Race':
        present = df[df[splitterColumn] ==  'black']
        absent = df[df[splitterColumn] == 'NotBlack'] 
        Present.append(present[columnNameNum])
        Absent.append(absent[columnNameNum])
    elif splitterColumn == 'Hisp':
        present = df[df[splitterColumn] ==  'hispanic']
        absent = df[df[splitterColumn] == 'NotHispanic'] 
        Present.append(present[columnNameNum])
        Absent.append(absent[columnNameNum])
        absent = df[df[splitterColumn]] 
    elif splitterColumn == 'MaritalStatus':
        present = df[df[splitterColumn] ==  'Married']
        absent = df[df[splitterColumn] == 'NotMarried'] 
        Present.append(present[columnNameNum])
        Absent.append(absent[columnNameNum])
    elif splitterColumn == 'Nodeg':
        present = df[df[splitterColumn] ==  1]
        absent = df[df[splitterColumn] == 0] 
        Present.append(present[columnNameNum])
        Absent.append(absent[columnNameNum])
        
    return round((abs(np.mean(Present) - np.mean(Absent))/df[columnNameNum].std()), 4), present[splitterColumn].value_counts(), absent[splitterColumn].value_counts()
    

effectSizer(df, splitterColumn, columnNameNum)
print(splitterColumn,"'s effect size on", columnNameNum, "is", effectSizer(df, splitterColumn, columnNameNum)) 

# 

# Effect sizes comparison

### For Earnings_1978, MaritalStatus has the greatest effect size, 0.5553 followed by Nodeg with 0.3329 then Race with 0.3177 and finally Hisp with just 0.1620. 
### The values were obtained by using all the columns in columnBinCategorical for the splitterColumn while maintaining columnNameNum = 'Earnings_1978'. 
```
